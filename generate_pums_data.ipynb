```python
from folktables import ACSDataSource, ACSIncome
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
data_source = ACSDataSource(survey_year='2021', horizon='1-Year', survey='person')
```

```python
original = data_source.get_data(density = 1, download=True)
```

```python
col = list(original.columns)
```
